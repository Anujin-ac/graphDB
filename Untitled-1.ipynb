````python
import os
import re
import json
import sys
import requests
import fitz

import numpy as np
import faiss

# PyTorch may fail to load on some Windows installations because of native DLL issues.
try:
    sys.path.insert(0, r"C:\torch_cpu")
    from sentence_transformers import SentenceTransformer
except (ImportError, OSError) as error:
    SentenceTransformer = None
    print(f"SentenceTransformer unavailable: {error}")

PDF_PATH = r"C:\Users\anuji\Downloads\infografik_medeelel_2026_on_6_sar_baganuur_2026-07.pdf"

doc = fitz.open(PDF_PATH)

print("Нийт хуудас:", len(doc))
````

````python
pages = []

for page_number, page in enumerate(doc):

    text = page.get_text("text")

    pages.append({
        "page": page_number + 1,
        "text": text
    })

print("PDF text extraction completed")

print(pages[0]["text"][:2000])
````

````python
def clean_text(text):

    text = text.replace("\n", " ")

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()

for page in pages:
    page["clean_text"] = clean_text(
        page["text"]
    )

print(pages[0]["clean_text"][:1000])
````

````python
def chunk_text(
    text,
    chunk_size=502
):

    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size 

    return chunks

document_id = "tt3"

all_chunks = []

chunk_counter = 0

for page in pages:

    page_chunks = chunk_text(
        page["clean_text"],
        chunk_size=502
    )

    for chunk in page_chunks:

        all_chunks.append({
            "chunk_id": f"chunk_{chunk_counter}",
            "document_id": document_id,
            "page": page["page"],
            "text": chunk
        })

        chunk_counter += 1

print("Нийт chunk:", len(all_chunks))
````

````python
# Store the generated chunks in FAISS VectorDB
import hashlib

VECTOR_DB_DIR = r"C:\Users\anuji\Downloads\baganuur_vector_db"
os.makedirs(VECTOR_DB_DIR, exist_ok=True)

embedding_dimension = 384

def create_lightweight_embeddings(texts, dimension=embedding_dimension):
    embeddings = []

    for text in texts:
        vector = np.zeros(dimension, dtype="float32")
        words = re.findall(r"\w+", text.lower(), flags=re.UNICODE)

        for word in words:
            digest = hashlib.sha256(word.encode("utf-8")).digest()
            position = int.from_bytes(digest[:4], "little") % dimension
            vector[position] += 1.0

        norm = np.linalg.norm(vector)
        if norm > 0:
            vector /= norm

        embeddings.append(vector)

    return np.asarray(embeddings, dtype="float32")

chunk_texts = [chunk["text"] for chunk in all_chunks]
chunk_embeddings = create_lightweight_embeddings(chunk_texts)

vector_index = faiss.IndexFlatIP(embedding_dimension)
vector_index.add(chunk_embeddings)

faiss.write_index(
    vector_index,
    os.path.join(VECTOR_DB_DIR, "chunks.faiss")
)

with open(
    os.path.join(VECTOR_DB_DIR, "chunks_metadata.json"),
    "w",
    encoding="utf-8"
) as metadata_file:
    json.dump(all_chunks, metadata_file, ensure_ascii=False, indent=2)

print(f"FAISS VectorDB-д {vector_index.ntotal} chunk хадгаллаа.")
print(f"Хадгалсан хавтас: {VECTOR_DB_DIR}")
````

````python
def escape_sparql_string(text):

    return (
        text
        .replace("\\", "\\\\")
        .replace('"', '\\"')
        .replace("\n", "\\n")
        .replace("\r", "")
    )
````

````python
def insert_chunks_to_graphdb():
    inserted_count = 0

    for chunk in all_chunks:
        chunk_id = chunk["chunk_id"]
        page = chunk["page"]
        text = escape_sparql_string(chunk["text"])

        query = f"""
        PREFIX ex: <{PREFIX}>

        INSERT DATA {{
            ex:{document_id}
                ex:hasChunk ex:{chunk_id} .

            ex:{chunk_id}
                a ex:Chunk ;
                ex:page {page} ;
                ex:text "{text}" .
        }}
        """

        try:
            response = requests.post(
                GRAPHDB_UPDATE_URL,
                data=query.encode("utf-8"),
                headers={
                    "Content-Type": "application/sparql-update; charset=utf-8",
                    "Accept": "application/json"
                },
                timeout=30
            )
            response.raise_for_status()
            inserted_count += 1

        except requests.exceptions.RequestException as error:
            print(f"Chunk insert failed: {chunk_id}")
            if "response" in locals():
                print(response.status_code, response.text[:500])
            print(error)
            return False

    print(f"GraphDB-д {inserted_count} chunk амжилттай орууллаа.")
    return inserted_count == len(all_chunks)


insert_chunks_to_graphdb()
````

````python
import requests
import json
import re

import getpass
import time

GOOGLE_AI_MODEL = (
    os.getenv("GOOGLE_AI_MODEL")
    or "gemini-3.5-flash-lite"
)
GOOGLE_API_KEY = (
    globals().get("GOOGLE_API_KEY")
    or os.getenv("GOOGLE_API_KEY")
    or getpass.getpass("Google AI API key: ")
)
if not GOOGLE_API_KEY:
    raise ValueError(
        "GOOGLE_API_KEY is required to call the Google AI API."
    )

GOOGLE_AI_URL = (
    "https://generativelanguage.googleapis.com/v1beta/models/"
    f"{GOOGLE_AI_MODEL}:generateContent"
)


def google_ai_generate(prompt, response_mime_type=None):
    generation_config = {"temperature": 0}
    if response_mime_type:
        generation_config["responseMimeType"] = response_mime_type

    request_body = {
        "contents": [{"parts": [{"text": prompt}]}],
        "generationConfig": generation_config
    }

    for attempt in range(3):
        try:
            response = requests.post(
                GOOGLE_AI_URL,
                params={"key": GOOGLE_API_KEY},
                json=request_body,
                timeout=(20, 300)
            )
            break
        except (
            requests.exceptions.Timeout,
            requests.exceptions.ConnectionError
        ) as error:
            if attempt == 2:
                raise RuntimeError(
                    "Google AI request timed out after 3 attempts. "
                    "Check your internet connection or try a shorter prompt."
                ) from error
            time.sleep(2 ** attempt)

    try:
        response.raise_for_status()
    except requests.exceptions.HTTPError as error:
        raise RuntimeError(
            f"Google AI API error for model '{GOOGLE_AI_MODEL}': "
            f"{response.text[:500]}"
        ) from error

    candidates = response.json().get("candidates", [])
    if not candidates:
        raise RuntimeError("Google AI returned no candidates.")
    return candidates[0]["content"]["parts"][0]["text"].strip()
````

````python
def make_uri(text):
    text = text.strip()

    text = re.sub(
        r"[^\w\s-]",
        "",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        "_",
        text
    )

    return text
````

````python
def extract_knowledge(text):

    prompt = f"""
Чи Монгол хэл дээрх "Багануур дүүргийн эдийн засаг, нийгмийн байдал" инфографикийн 
мэдээллээс entity болон relationship гаргах үүрэгтэй Knowledge Graph систем.

Дараах текстийг уншиж, Багануур дүүргийн эдийн засаг, нийгмийн үзүүлэлт, байгууллага,
хүн ам, салбар, бодлого, нөхцөл байдалтай холбоотой entity болон relationship-ийг 
гадаадын төрлөөр бус Монгол хэл дээрх утга учиртай нэршилтэйгээр гарга.

Entity type:
- District
- Organization
- Person
- Population
- Sector
- Indicator
- Value
- Policy
- SocialIssue
- Concept

Relationship:
- hasValue
- hasIndicator
- hasPopulation
- operatesIn
- locatedIn
- belongsTo
- affects
- supports
- promotes
- measures
- develops
- addresses
- contributesTo
- comparesTo
- defines

Зөвхөн JSON буцаа. Хариулт нь энэ форматтай байх ёстой:

{{
  "entities": [
    {{
      "name": "entity нэр",
      "type": "entity type"
    }}
  ],
  "relationships": [
    {{
      "subject": "entity",
      "predicate": "relationship",
      "object": "entity"
    }}
  ]
}}

Заавар:
- Багануур дүүрэг, дүүргийн Засаг дарга, орон нутгийн удирдлага, төрийн болон хувийн байгууллагууд,
  хүн ам, аж ахуйн нэгж, салбар, эдийн засгийн үзүүлэлт, нийгмийн асуудал зэрэгтэй холбоотой
  мэдээллийг авч, нэр томьёо нь эх сурвалжийн тексттэй нийцсэн байх.
- Хэрэв текстэд тоон утга, хувь, хэмжээ, түвшин байвал "Value" entity болгон гаргаж болно.
- Relationship-ийг зөвхөн өмнөх жагсаалтаас сонгоно.
- Өөр ямар ч текст, тайлбар, markdown, code fence, note бүү нэм.
- Зөвхөн JSON объект л буцаа.

Текст:

{text}
"""

    try:
        return google_ai_generate(
            prompt,
            response_mime_type="application/json"
        )
    except requests.exceptions.RequestException as error:
        print("Google AI API error:", error)
        return None
    except (KeyError, IndexError, RuntimeError) as error:
        print("Google AI response error:", error)
        return None
````

````python
def parse_knowledge(raw):
    if not raw:
        print("No knowledge returned by the AI model; skipping this chunk.")
        return {"entities": [], "relationships": []}

    cleaned = raw.strip()
    cleaned = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        cleaned,
        flags=re.IGNORECASE
    )

    try:
        data = json.loads(cleaned)
    except json.JSONDecodeError as error:
        print("JSON error:", error)
        print("Raw response:")
        print(raw)
        return {"entities": [], "relationships": []}

    if not isinstance(data, dict):
        print("Knowledge payload was not a JSON object; skipping this chunk.")
        return {"entities": [], "relationships": []}

    entities = data.get("entities") or []
    relationships = data.get("relationships") or []

    if not isinstance(entities, list):
        entities = []
    if not isinstance(relationships, list):
        relationships = []

    return {
        "entities": entities,
        "relationships": relationships
    }
````

````python
ALLOWED_RELATIONSHIPS = {
    "hasValue",
    "hasIndicator",
    "hasPopulation",
    "operatesIn",
    "locatedIn",
    "belongsTo",
    "affects",
    "supports",
    "promotes",
    "measures",
    "develops",
    "addresses",
    "contributesTo",
    "comparesTo",
    "defines"
}


def entity_uri(name):
    value = make_uri(str(name))
    return f"{PREFIX}{value or 'UnnamedEntity'}"


def insert_knowledge_graph(knowledge_data, source_chunk):
    if knowledge_data is None:
        raise ValueError("knowledge_data is empty")

    entities = {
        str(entity.get("name", "")).strip(): entity
        for entity in (knowledge_data.get("entities") or [])
        if str(entity.get("name", "")).strip()
    }

    for relationship in (knowledge_data.get("relationships") or []):
        for key in ("subject", "object"):
            name = str(relationship.get(key, "")).strip()
            if name and name not in entities:
                entities[name] = {"name": name, "type": "Concept"}

    triples = []
    chunk_uri = f"{PREFIX}{source_chunk['chunk_id']}"

    for name, entity in entities.items():
        node_uri = entity_uri(name)
        entity_type = make_uri(entity.get("type", "Concept")) or "Concept"
        triples.extend([
            f"<{node_uri}> a ex:Entity, ex:{entity_type} ;",
            f"    ex:name \"{escape_sparql_string(name)}\" ;",
            f"    ex:entityType \"{escape_sparql_string(entity.get('type', 'Concept'))}\" ;",
            f"    ex:mentionedIn <{chunk_uri}> ."
        ])

    edge_count = 0
    for relationship in (knowledge_data.get("relationships") or []):
        subject = str(relationship.get("subject", "")).strip()
        predicate = str(relationship.get("predicate", "")).strip()
        object_name = str(relationship.get("object", "")).strip()

        if not subject or not predicate or not object_name:
            continue
        if predicate not in ALLOWED_RELATIONSHIPS:
            print(f"Skipped unknown relationship: {predicate}")
            continue

        triples.append(
            f"<{entity_uri(subject)}> ex:{predicate} <{entity_uri(object_name)}> ."
        )
        edge_count += 1

    if not triples:
        print(f"No valid triples to insert for chunk {source_chunk.get('chunk_id')}; skipping.")
        return {"nodes": 0, "edges": 0}

    update = "PREFIX ex: <{}>\n\nINSERT DATA {{\n{}\n}}".format(
        PREFIX,
        "\n".join(triples)
    )
    response = requests.post(
        GRAPHDB_UPDATE_URL,
        data=update.encode("utf-8"),
        headers={"Content-Type": "application/sparql-update; charset=utf-8"},
        timeout=30
    )
    response.raise_for_status()

    print(
        f"GraphDB-д {len(entities)} node, {edge_count} edge орууллаа."
    )
    return {"nodes": len(entities), "edges": edge_count}


insert_knowledge_graph(knowledge, test_chunk)
````

````python
def build_graph_from_all_chunks():
    total_nodes = 0
    total_edges = 0
    failed_chunks = []

    for chunk_number, source_chunk in enumerate(all_chunks, start=1):
        print(f"Processing chunk {chunk_number}/{len(all_chunks)}...")
        raw = extract_knowledge(source_chunk["text"])
        parsed = parse_knowledge(raw)

        if not parsed:
            failed_chunks.append(source_chunk["chunk_id"])
            continue

        try:
            counts = insert_knowledge_graph(parsed, source_chunk)
            total_nodes += counts["nodes"]
            total_edges += counts["edges"]
        except requests.exceptions.RequestException as error:
            print(f"GraphDB insert failed for {source_chunk['chunk_id']}: {error}")
            failed_chunks.append(source_chunk["chunk_id"])

    print(f"Нийт боловсруулсан node: {total_nodes}")
    print(f"Нийт боловсруулсан edge: {total_edges}")
    print(f"Алдаатай chunk: {failed_chunks}")

    return {
        "nodes": total_nodes,
        "edges": total_edges,
        "failed_chunks": failed_chunks
    }


graph_result = build_graph_from_all_chunks()
````

````python
from concurrent.futures import ThreadPoolExecutor


def fetch_chunks_from_graphdb():
    query = f"""
    PREFIX ex: <{PREFIX}>

    SELECT ?chunk ?page ?text
    WHERE {{
        ex:{document_id} ex:hasChunk ?chunk .
        ?chunk ex:page ?page ;
               ex:text ?text .
    }}
    ORDER BY ?page
    """

    response = requests.get(
        GRAPHDB_QUERY_URL,
        params={"query": query},
        headers={"Accept": "application/sparql-results+json"},
        timeout=30
    )
    response.raise_for_status()

    chunks = []
    for binding in response.json()["results"]["bindings"]:
        chunks.append({
            "chunk_id": binding["chunk"]["value"].rsplit("/", 1)[-1],
            "page": int(binding["page"]["value"]),
            "text": binding["text"]["value"]
        })
    return chunks


def extract_question_terms(question):
    return set(
        term for term in re.findall(r"[\wӨөҮүЁё-]+", question.lower(), flags=re.UNICODE)
        if len(term) > 1
    )


def calculate_relevance(question, text):
    question_terms = extract_question_terms(question)
    text_terms = extract_question_terms(text)
    if not question_terms:
        return 0.0
    return len(question_terms & text_terms) / len(question_terms)


def retrieve_from_faiss(question, top_k=3):
    if vector_index.ntotal == 0:
        raise RuntimeError("FAISS VectorDB-д chunk олдсонгүй.")

    question_embedding = create_lightweight_embeddings([question])
    similarities, indices = vector_index.search(
        question_embedding,
        min(top_k, vector_index.ntotal)
    )

    with open(
        os.path.join(VECTOR_DB_DIR, "chunks_metadata.json"),
        "r",
        encoding="utf-8"
    ) as metadata_file:
        vector_chunks = json.load(metadata_file)

    selected = []
    for similarity, index_number in zip(similarities[0], indices[0]):
        chunk = vector_chunks[int(index_number)].copy()
        chunk["score"] = float(similarity)
        chunk["source"] = "FAISS VectorDB"
        selected.append(chunk)
    return selected


def retrieve_from_graphdb(question, top_k=3):
    graph_chunks = fetch_chunks_from_graphdb()
    if not graph_chunks:
        return []

    for chunk in graph_chunks:
        chunk["score"] = calculate_relevance(question, chunk["text"])
        chunk["source"] = "GraphDB"

    return sorted(
        graph_chunks,
        key=lambda chunk: chunk["score"],
        reverse=True
    )[:top_k]


def compare_retrieval(question, top_k=3):
    with ThreadPoolExecutor(max_workers=2) as executor:
        faiss_future = executor.submit(retrieve_from_faiss, question, top_k)
        graph_future = executor.submit(retrieve_from_graphdb, question, top_k)
        faiss_chunks = faiss_future.result()
        graph_chunks = graph_future.result()

    faiss_score = max(
        (chunk["score"] for chunk in faiss_chunks),
        default=0.0
    )
    graph_score = max(
        (chunk["score"] for chunk in graph_chunks),
        default=0.0
    )

    if graph_score > faiss_score:
        winner = "GraphDB"
        selected_chunks = graph_chunks
    else:
        winner = "FAISS VectorDB"
        selected_chunks = faiss_chunks

    print("\n===== Retrieval comparison =====")
    print(f"FAISS VectorDB: {len(faiss_chunks)} chunk, score={faiss_score:.4f}")
    for chunk in faiss_chunks:
        print(f"  - {chunk['chunk_id']} (page {chunk['page']}, score={chunk['score']:.4f})")
    print(f"GraphDB: {len(graph_chunks)} chunk, score={graph_score:.4f}")
    for chunk in graph_chunks:
        print(f"  - {chunk['chunk_id']} (page {chunk['page']}, score={chunk['score']:.4f})")
    print(f"Илүү оновчтой эх сурвалж: {winner}")

    return {
        "winner": winner,
        "faiss_chunks": faiss_chunks,
        "graph_chunks": graph_chunks,
        "selected_chunks": selected_chunks,
        "faiss_score": faiss_score,
        "graph_score": graph_score
    }


def ask_graphdb(question, top_k=3):
    comparison = compare_retrieval(question, top_k=top_k)
    relevant_chunks = comparison["selected_chunks"]
    if not relevant_chunks:
        raise RuntimeError("FAISS болон GraphDB-ээс тохирох chunk олдсонгүй.")

    context = "\n\n".join(
        f"[Хуудас {chunk['page']}, {chunk['chunk_id']}]\n{chunk['text']}"
        for chunk in relevant_chunks
    )

    prompt = f"""
Та Багануур дүүргийн эдийн засаг, нийгмийн байдлын асуултад хариулагч туслах.

Дараах хамгийн оновчтой retrieval эх сурвалжаас авсан мэдээлэлд тулгуурлан
асуултад Монгол хэлээр товч бөгөөд тодорхой хариул. Эх сурвалжид байхгүй
мэдээллийг зохиож болохгүй. Хариултын төгсгөлд ашигласан хуудасны дугаарыг дурд.

Сонгогдсон эх сурвалж: {comparison['winner']}
Эх сурвалж:
{context}

Асуулт:
{question}
"""

    answer = google_ai_generate(prompt)
    print("\n===== Final answer =====")
    print("Асуулт:", question)
    print("Хариулт:", answer)
    print(f"Хариулт сонгосон эх сурвалж: {comparison['winner']}")
    return answer
````

````python
def fetch_entities_from_graphdb():
    query = f"""
    PREFIX ex: <{PREFIX}>

    SELECT ?entity ?name ?entity_type
    WHERE {{
        ?entity a ex:Entity ;
                ex:name ?name ;
                ex:entityType ?entity_type .
    }}
    """

    response = requests.get(
        GRAPHDB_QUERY_URL,
        params={"query": query},
        headers={"Accept": "application/sparql-results+json"},
        timeout=30
    )
    response.raise_for_status()

    return [
        {
            "uri": item["entity"]["value"],
            "name": item["name"]["value"],
            "type": item["entity_type"]["value"]
        }
        for item in response.json()["results"]["bindings"]
    ]


def detect_question_entities(question):
    question_lower = question.lower()
    entities = []

    for entity in fetch_entities_from_graphdb():
        if entity["name"].lower() in question_lower:
            entities.append(entity)

    return entities
````

````python
def fetch_entity_chunks(question):
    entities = detect_question_entities(question)

    if not entities:
        return []

    entity_values = " ".join(
        f"<{entity['uri']}>" for entity in entities
    )

    query = f"""
    PREFIX ex: <{PREFIX}>

    SELECT DISTINCT ?chunk ?page ?text ?entity ?entity_name
    WHERE {{
        VALUES ?entity {{ {entity_values} }}

        ?entity ex:mentionedIn ?chunk ;
                ex:name ?entity_name .

        ?chunk a ex:Chunk ;
               ex:page ?page ;
               ex:text ?text .
    }}
    ORDER BY ?page
    """

    response = requests.get(
        GRAPHDB_QUERY_URL,
        params={"query": query},
        headers={"Accept": "application/sparql-results+json"},
        timeout=30
    )
    response.raise_for_status()

    chunks = {}

    for item in response.json()["results"]["bindings"]:
        chunk_id = item["chunk"]["value"].rsplit("/", 1)[-1]

        if chunk_id not in chunks:
            chunks[chunk_id] = {
                "chunk_id": chunk_id,
                "page": int(item["page"]["value"]),
                "text": item["text"]["value"],
                "matched_entities": []
            }

        chunks[chunk_id]["matched_entities"].append(
            item["entity_name"]["value"]
        )

    return list(chunks.values())
````

````python
def fetch_related_chunks(question):
    direct_chunks = fetch_entity_chunks(question)

    if not direct_chunks:
        return []

    chunk_values = " ".join(
        f"<{PREFIX}{chunk['chunk_id']}>"
        for chunk in direct_chunks
    )

    query = f"""
    PREFIX ex: <{PREFIX}>

    SELECT DISTINCT ?chunk ?page ?text ?relatedness
    WHERE {{
        VALUES ?source_chunk {{ {chunk_values} }}

        ?source_chunk ex:relatedChunk ?chunk ;
                      ex:sharedEntityCount ?relatedness .

        ?chunk a ex:Chunk ;
               ex:page ?page ;
               ex:text ?text .
    }}
    ORDER BY DESC(?relatedness)
    """

    response = requests.get(
        GRAPHDB_QUERY_URL,
        params={"query": query},
        headers={"Accept": "application/sparql-results+json"},
        timeout=30
    )
    response.raise_for_status()

    return [
        {
            "chunk_id": item["chunk"]["value"].rsplit("/", 1)[-1],
            "page": int(item["page"]["value"]),
            "text": item["text"]["value"],
            "entity_score": float(item["relatedness"]["value"]),
            "source": "GraphDB entity layer"
        }
        for item in response.json()["results"]["bindings"]
    ]
````

````python
def retrieve_from_graphdb(question, top_k=3):
    direct_chunks = fetch_entity_chunks(question)
    related_chunks = fetch_related_chunks(question)

    results = {}

    for chunk in direct_chunks:
        chunk["score"] = 1.0
        chunk["source"] = "GraphDB direct entity"
        results[chunk["chunk_id"]] = chunk

    for chunk in related_chunks:
        chunk_id = chunk["chunk_id"]

        if chunk_id not in results:
            chunk["score"] = min(
                0.9,
                0.5 + 0.1 * chunk["entity_score"]
            )
            results[chunk_id] = chunk

    return sorted(
        results.values(),
        key=lambda item: item["score"],
        reverse=True
    )[:top_k]
````
